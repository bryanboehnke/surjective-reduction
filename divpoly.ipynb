If $P= (x,y)$, then 

$$x(3P) = \dfrac{\phi_3(P)}{\psi_3(P)^2},$$
<!-- = \dfrac{x \psi_3(P)^2 - \psi_2(P) \psi_4(P)}{\psi_3(P)^2} -->

where

$$\psi_3(x,y) = 3x^4 + b_2 x^3 + 3b_4 x^2 + 3b_6 x + b_8$$

and 

\begin{align*}
\phi_3(x,y) &= x \psi_3(P)^2 - \psi_2(P) \psi_4(P) \\
& = x \psi_3(P)^2 - \psi_2(P)^2 \left(2x^6 + b_2^5+ 5b_4^4+ 10b_6x^3+10b_8x^2 + (b_2b_8-b_4b_6)x + (b_4b_8-b_6^2) \right) \\
&=  x \left(3x^4 + b_2 x^3 + 3b_4 x^2 + 3b_6 x + b_8\right)^2 - \left( 4x^3 + b_2x^2+2b_4x + b_6\right) \left(2x^6 + b_2^5+ 5b_4^4+ 10b_6x^3+10b_8x^2 + (b_2b_8-b_4b_6)x + (b_4b_8-b_6^2) \right) \\
\end{align*}

```python
a1 = E[0]
a2 = E[1]
a3 = E[2]
a4 = E[3]
a6 = E[4]

b2 = a1**2 + 4*a2
b4 = 2*a4 + a1*a3
b6 = a3**2 + 4*a6
b8 = a1**2 * a6 + 4*a2*a6 - a1*a3*a4 + a2*a3**2 - a4**2


x = Symbol('x')

# \psi_3 = 3x^4 + b2 x^3 + 3b4 x^2 + 3b6 x + b8
psi3 = Poly([3,b2,3*b4,3*b6, b8], x)

psi2sq = Poly([4,b2,2*b4,b6], x)
psi4bypsi2 = Poly([2,b2,5*b4,10*b6,10*b8,(b2*b8-b4*b6),(b4*b8-b6**2)], x)
phi3 = x*psi3**2-psi2sq*psi4bypsi2

f = Poly([1,0,-189,-999],x)
```
